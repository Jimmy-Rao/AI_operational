```python
import pandas as pd

# 加载数据集并显示数据集的前五行 1分
data = pd.read_csv('2.1.1_cleaned_data.csv')
print("数据集的前五行:")
print(data.head())

# 显示每一列的数据类型
print(data.dtypes)

# 检查清洗后的缺失值
print("\n检查清洗后的缺失值:")
print(data.isnull().sum())  
data =data.dropna()  # 删除缺失值

# 将 'horsepower' 列转换为数值类型，并处理转换中的异常值 1分
data['horsepower'] = pd.to_numeric(data['horsepower'], errors='coerce')
data = data.dropna(subset=['horsepower']) # 删除转换后为 NaN 的行

# 显示每一列的数据类型
print(data.horsepower.dtypes)
```
