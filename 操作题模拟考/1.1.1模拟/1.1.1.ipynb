```python
import pandas as pd
import numpy as np

# 读取数据
data = pd.read_csv('patient_data.csv')

# 1. 统计住院天数超过7天的患者数量及其占比
# 创建新列'RiskLevel'，根据住院天数判断风险等级
data['RiskLevel'] = np.where(data['DaysInHospital'] > 7, '高风险患者', '低风险患者')
# 统计不同风险等级的患者数量
risk_counts = data['RiskLevel'].value_counts()
# 计算高风险患者占比
high_risk_percentage = risk_counts['高风险患者'] / risk_counts.sum()
# 低风险患者占比
low_risk_percentage = risk_counts['低风险患者'] / risk_counts.sum()

# 输出结果
print("高风险患者数量:", risk_counts['高风险患者'])
print("低风险患者数量:", risk_counts['低风险患者'])
print("高风险患者占比:", high_risk_percentage)
print("低风险患者占比:", low_risk_percentage)

# 2. 统计不同BMI区间中高风险患者的比例和患者数
# 定义BMI区间和标签
bmi_bins = [0, 18.5, 24, 28, np.inf]
bmi_labels = ['偏瘦', '正常', '超重', '肥胖']
# 根据BMI值分配BMI区间标签
data['BMIRange'] = pd.cut(data['BMI'], bins=bmi_bins, labels=bmi_labels, right=False)
# 计算每个BMI区间中高风险患者的比例
bmi_risk_percentage = data.groupby('BMIRange')['RiskLevel'].apply(lambda x: (x == '高风险患者').mean())
# 统计每个BMI区间的患者数量
bmi_percentage_count = data.groupby('BMIRange').size()

# 输出结果
print("BMI区间中高风险患者的比例和患者数:")
print(bmi_risk_percentage)    #高风险患者的比例
print(bmi_percentage_count)    #高风险患者的患者数

# 2. 统计不同BMI区间中高风险患者的比例和患者数
# 定义BMI区间和标签
age_bins = [0, 26, 36, 46, 56, 66, np.inf]
age_labels = ['≤25岁', '26岁-35岁', '36岁-45岁', '46岁-55岁', '56岁-65岁', '＞65岁']
# 根据年龄值分配年龄区间标签
data['AgeRange'] = pd.cut(data['Age'], bins=age_bins, labels=age_labels, right=False)
# 计算每个年龄区间中高风险患者的比例
age_risk_percentage = data.groupby('AgeRange')['RiskLevel'].apply(lambda x: (x == '高风险患者').mean())
# 统计每个年龄区间的患者数量
age_percentage_count = data.groupby('AgeRange').size()

# 输出结果
print("年龄区间中高风险患者的比例和患者数:")
print(age_risk_percentage)   #高风险患者的比例
print(age_percentage_count)   #高风险患者的患者数
```
